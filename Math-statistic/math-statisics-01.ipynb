## Percentage
A hotel has 200 rooms.

On Monday, 146 rooms were occupied.

Calculate:

- Occupancy percentage.
- Percentage of rooms that were empty.
- If occupancy increases to 160 rooms on Tuesday, by how many percentage points did occupancy increase?
- By what percentage did the number of occupied rooms increase from Monday to Tuesday?

```python
hotel_room = 200
monday_occup = 146
occupancy_pct_monday = monday_occup/hotel_room * 100 
empty_room_pct = (hotel_room-monday_occup)/hotel_room*100
tuesday_occup = 160
pct_occupancy_monday_tuesday = (tuesday_occup-monday_occup) / monday_occup * 100
pct_points_occup_increase = tuesday_occup/hotel_room * 100 - occupancy_pct_monday

print(f"Occupancy percentage: {occupancy_pct_monday:.0f}%")
print(f"Percentage of rooms that were empty: {empty_room_pct:.0f}%")
print(f"Tuesday occupancy was increased by: {pct_points_occup_increase:.0f} percentage points")
print(f"Percentage of occupied rooms increase from Monday to Tuesday: {pct_occupancy_monday_tuesday:.2f}%")
```

## Calculate the mean and median of 12000, 13500, 14000, 14500, 15000, 15500, 40000, and explain which better represents a typical day.

```python
data = [12000, 13500, 14000, 14500, 15000, 15500, 40000]
total = 0 
for i in data:
    total += i 
# mean and median of the data
data_mean = total / len(data)
if len(data)%2 == 0: 
    data_median = data[len(data) // 2]
else:
    data_median = (data[len(data)//2-1] + data[len(data)//2])/2
data_mean, data_median
```

- The median value better represent a typical day.

# Task 3: Range and IQR

```python
dataset = [10, 12, 13, 14, 15, 16, 18, 40]

# min value 
dataset_min = min(dataset)
# max value
dataset_max = max(dataset) 
# Range 
dataset_range = dataset_max - dataset_min
# frist and third percentile 
dataset_length = len(dataset)
first_halve = dataset[:dataset_length//2]
second_halve = dataset[dataset_length//2:]
Q1 = (first_halve[len(first_halve)//2-1] + first_halve[len(first_halve)//2])/2
Q3 = (second_halve[len(second_halve)//2-1] +second_halve[len(second_halve)//2])/2
# Interquartile range - IQR 
IQR = Q3 - Q1 
print(f"Min value is: {dataset_min}\nMax value is: {dataset_max}\nRange is {dataset_range}\nQ1={Q1}, Q3={Q3} and IQR={IQR}")
```

Then think about whether 40 looks unusual compared with the rest of the data.

```python
lower_bound = Q1 - 1.5 * IQR 
upper_bound = Q3 + 1.5 * IQR
lower_bound, upper_bound 
```

40 Looks very unusual compared to the rest of data, looks like an outlier

# Task 4 — Basic Probability

A hotel received 1,000 bookings:

Booking status	Number
Completed	720
Cancelled	200
No-show	80

If one booking is selected randomly, calculate:

P(Completed)
P(Cancelled)
P(No-show)
P(Not cancelled)
P(Cancelled or No-show)

Write the answers both as decimals and percentages.

```python
completed_bookings = 720
cancelled_bookings = 200 
no_show = 80
total = completed_bookings+cancelled_bookings+no_show
# Probability for selecting randomly completed booking 
P_completed = completed_bookings / total
print(f"Probability for selecting randomly completed booking: {P_completed:.2f}, {P_completed* 100:.2f}%")
# Probability for selecting randomly cancelled booking 
P_cancelled = cancelled_bookings / total 
print(f"Probability for selecting randomly cancelled booking: {P_cancelled:.2f}, {P_cancelled* 100:.2f}%")
# Probability for selecting randomly no-show booking 
P_no_show = no_show/total 
print(f"Probability for selecting randomly no-show booking: {P_no_show:.2f}, {P_no_show* 100:.2f}%")
# Probability for selecting randomly not cancelled 
P_not_cancelled = 1 - P_cancelled
print(f"Probability for selecting randomly not cancelled : {P_not_cancelled:.2f}, {P_not_cancelled* 100:.2f}%")
# Probability for selecting randomly cancelled OR no-show booking 
P_cancelled_no_show = P_cancelled + P_no_show
print(f"Probability for selecting randomly cancelled OR no-show booking : {P_cancelled_no_show:.2f}, {P_cancelled_no_show* 100:.2f}%")
```
