# Introduction to the One-Sample t-Test
Until now, our z-tests assumed the population standard deviation σ was known.

In real problems, σ is often unknown. Then we estimate variability using the sample standard deviation s, and use a t-test.

Use this scenario:

A hotel claims its average daily revenue is:

μ0=20,000

You take a sample of:

n=16

and observe:

with sample standard deviation:

s=2,400

We want to test whether the true mean is different from £20,000.

H1:

two-tailed

### Part B — Standard Error

Because σ is unknown, use:

SE = s / sqrt(n)

Calculate the standard error.

### Part C — t-statistic

Now calculate:

t = (x - m0) / SE 

Then interpret the result in words:

The sample mean is ___ standard errors above/below the hypothesized mean.

### Part D — Degrees of freedom

For a one-sample t-test:

df=n−1

Calculate df.

Then answer:

Why do you think the t-distribution needs this extra value df, while the standard normal z-distribution does not?


```python
import numpy as np 
import scipy.stats as st 

m0 = 20000
n = 16
x_bar = 21200
s = 2400

print("H0: m0 == 20000, true mean is equal to 20000")
print("H1: m0 == 20000, true mean is different from 20000")
print("This is two-tailed test.")
SE = s / np.sqrt(n) 
print(f"Standard error is {SE}")
t = (x_bar - m0) / SE 
print(f"The sample mean is {t} standard error above the hypothesized mean.")
df = n - 1 
print("Because the population standard deviation is unknown, we estimate it using the sample standard deviation s.\
That introduces extra uncertaintly. The t-distribution uses degrees of freedom to account for how much information we have when estimating that variability.")
```

### Part E: calculate the p-value from t=2 with df=15 using the t-distribution.

```python
p_value = 2 * (1 - st.t.cdf(abs(t),df))
print(f"p_value is {p_value}")
print(f"Since p_value is greater than significance level , {p_value} > 0.05, we fail to reject H0.")
print("The sample mean is 2 standard deviation above 20000, but with only n = 16 and unknown population standard deviation, that \
is not quite extreme enough to be statistically significant at the 5% level.")
```

# Independent two-sample t-test
A hotel tests two pricing strategies on different groups of days:

strategy_a = np.array([
    20100, 20500, 19800, 21000, 20700,

strategy_b = np.array([

    21300, 21900, 21100, 21700, 21500
])

The samples are independent: observations in Strategy A are not matched with observations in Strategy B.

Use:

α=0.05

We want to test whether the two strategies have different true mean revenues.

Part A — Hypotheses

Write:

H0:

H1:

Also identify whether this is one-tailed or two-tailed.

Part B — Descriptive statistics

For each group calculate:

2. x_barA , x_barB
3. sA, sB
   
Then calculate the observed difference:

x_barA - x_barB

Interpret its sign.


### Part C — Standard error of the difference

For now we'll use Welch's independent t-test, which does not require us to assume equal population variances.

Calculate:

SExa-xb = sqrt(sA^2/nA + sB^2/nB)

Then calculate:

t = (xA-xB) / SE 

```python
strategy_a = np.array([
    20100, 20500, 19800, 21000, 20700,
    20200, 20900, 20400, 20600, 20000
])

strategy_b = np.array([
    21400, 21800, 20900, 22100, 21600,
    21300, 21900, 21100, 21700, 21500
])

alpha = 0.05
print(f"H0: {chr(956)}A == {chr(956)}B")
print(f"H1: {chr(956)}A != {chr(956)}B")
print("I would use two-tailed test, because I care about difference in either direction.")

nA = len(strategy_a)
nB = len(strategy_b)
sample_mean_A = strategy_a.mean()
sample_mean_B = strategy_b.mean()
s_A = np.std(strategy_a,ddof=1) 
s_B = np.std(strategy_b,ddof=1) 
print(f"Strategy A has size = {nA}, sample mean = {sample_mean_A} and standard deviation = {s_A}")
print(f"Strategy B has size = {nB}, sample mean = {sample_mean_B} and standard deviation = {s_B}")
observed_difference = sample_mean_A-sample_mean_B
print(f"Observed difference between sample mean A and sample mean B is {observed_difference}, minus means sample mean A is smaller than sample mean B.")
SE = np.sqrt(s_A**2/ nA + s_B**2/nB)
print(f"The estimated standard deviation of the difference between the two sample means is about £{SE}")
t = observed_difference / SE 
print(f"The observed difference between Strategy A and Strategy B is about {SE} standard errors below the difference we assumed by H0.")
df = (s_A**2/ nA + s_B**2/nB)**2 / ((s_A**2/nA)**2 / (nA-1) + (s_B**2/nB)**2/(nB-1))

p_value = 2 * (1 - st.t.cdf(abs(t), df))

print(f"P-value is {p_value}, which is less than alpha 0.05, therefore there is statistically significant evidence that the true mean revenues of Strategy A and \
Strategy B are different.")
print(f"Strategy B has larger sample mean.")
st.ttest_ind(strategy_a, strategy_b,equal_var=False)
```

# Paired t-test
Now we move from two independent groups to paired measurements.

A hotel measures average daily restaurant revenue for the same 8 days before and after a menu change:
before = np.array([
    8200, 7900, 8400, 8100,
    8000, 8300, 8150, 8050
])

after = np.array([
    8500, 8200, 8600, 8350,
    8250, 8550, 8400, 8300
])

alpha = 0.05
We want to test whether the menu change increased average revenue.

### Part A — Hypotheses

Define the paired difference as:

d=after−before

Then write:

H0:

H1:

in terms of the mean difference μd.

Also identify whether this is left-tailed, right-tailed, or two-tailed.

### Part B — Build the differences
Then find:

1. n
2. mean difference dˉ
3. sample standard deviation of the differences sd

### Part C — Standard error

Calculate:

SEd = sd / sqrt(n)

### Part D — t-statistic

Because under H0 the expected mean difference is 0:

t= d-0 / SEd

Also calculate:

df=n−1

Do Parts A–D only first. Do not use st.ttest_rel() yet.

```python
before = np.array([
    8200, 7900, 8400, 8100,
    8000, 8300, 8150, 8050
])

after = np.array([
    8500, 8200, 8600, 8350,
    8250, 8550, 8400, 8300
])

alpha = 0.05 

print(f"H0: {chr(956)}d <= 0")
print(f"H1: {chr(956)}d > 0")
print(f"I would use right-tailed test, because I care about the difference in right direction.")

differences = after - before
n = len(differences)
d = differences.mean()
sd = np.std(differences,ddof=1) 

print(f"length of differences is {n}, mean is {d}, and standard deviation is {sd}.")

SE = sd / np.sqrt(n) 
print(f"Standard error of the mean paired difference is about {SE}")
t = (d-0) / SE 
print(f"Observed difference between after and before changes in menu is about {t} standard errors above the null-hypothesized difference 0.")
df = n - 1 
p_value = 1 - st.t.cdf(t,df) 
print(f"p value is less than alpha {p_value} < 0.05, => there is statistically significant evidence that the mean paired difference in revenue after versus before the menu change is positive.")

st.ttest_rel(after, before,alternative='greater')
```
