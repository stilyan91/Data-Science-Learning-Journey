# Introduction to the One-Sample t-Test
Until now, our z-tests assumed the population standard deviation σ was known.

In real problems, σ is often unknown. Then we estimate variability using the sample standard deviation s, and use a t-test.

Use this scenario:

A hotel claims its average daily revenue is:

μ0=20,000

You take a sample of:

n=16

and observe:

xˉ=21,200

with sample standard deviation:

s=2,400

We want to test whether the true mean is different from £20,000.

### Part A — Hypotheses

H0: 

H1:

Then say whether this is:

left-tailed
right-tailed
two-tailed

### Part B — Standard Error

Because σ is unknown, use:

SE = s / sqrt(n)

Calculate the standard error.

### Part C — t-statistic

Now calculate:

t = (x - m0) / SE 

Then interpret the result in words:

The sample mean is ___ standard errors above/below the hypothesized mean.

### Part D — Degrees of freedom

For a one-sample t-test:

df=n−1

Calculate df.

Then answer:

Why do you think the t-distribution needs this extra value df, while the standard normal z-distribution does not?


```python
import numpy as np 
import scipy.stats as st 

m0 = 20000
n = 16
x_bar = 21200
s = 2400

print("H0: m0 == 20000, true mean is equal to 20000")
print("H1: m0 == 20000, true mean is different from 20000")
print("This is two-tailed test.")
SE = s / np.sqrt(n) 
print(f"Standard error is {SE}")
t = (x_bar - m0) / SE 
print(f"The sample mean is {t} standard error above the hypothesized mean.")
df = n - 1 
print("Because the population standard deviation is unknown, we estimate it using the sample standard deviation s.\
That introduces extra uncertaintly. The t-distribution uses degrees of freedom to account for how much information we have when estimating that variability.")
```

### Part E: calculate the p-value from t=2 with df=15 using the t-distribution.

```python
p_value = 2 * (1 - st.t.cdf(abs(t),df))
print(f"p_value is {p_value}")
print(f"Since p_value is greater than significance level , {p_value} > 0.05, we fail to reject H0.")
print("The sample mean is 2 standard deviation above 20000, but with only n = 16 and unknown population standard deviation, that \
is not quite extreme enough to be statistically significant at the 5% level.")
```

```python
revenues = np.array([
    21400, 19800, 22100, 20500,
    23000, 21700, 20200, 22500,
    20900, 21800, 21100, 22400
])
m0 = 20000 
print('H0: m0 == 20000, the average is equal to 20000.')
print('H1: m0 != 20000, the average is different from 20000.')
print("It is two-tailed test because we test if m0 is exactly 20000.")

n = len(revenues)
observed_mean = revenues.mean()
s = np.std(revenues,ddof=1)
SE = s / np.sqrt(n) 
t = (observed_mean-m0) / SE 
print(f"The sample is {t} standard deviations above the hypothesized mean.")
df = n - 1
p_value = 2 * (1 - st.t.cdf(abs(t), df))
alpha = 0.05
print(f"p value {p_value} is less than aplha(0.05), so we reject H0.")
print(f"The sample mean is {t} standard deviation above the 20000, but ")
print("There is statistically significant evidence that the true mean is different from 20000.")
st.ttest_1samp(revenues,20000)
```

# One-sample t-test + Confidence Interval
Use the same sample from Task 14
We are still testing:

and you already found that the result is statistically significant.

Now we want to estimate a plausible range for the true mean.

### Part A — Sample statistics

Calculate again:

- s

### Part B — Critical t-value

For a 95% confidence interval:

α=0.05

Because this is two-sided, find the critical t-value for:

1 - alpha / 2


Use the t-distribution and your df.

### Part C — Margin of Error

Calculate:

ME=t_critical⋅SE

Then calculate the confidence interval:

xˉ−ME

and

xˉ+ME

### Part D — Interpretation

1. What is your 95% confidence interval?
2. Does the interval contain £20,000?
3. How does that agree with your hypothesis-test result?
4. In your own words, what does the 95% confidence interval mean?

```python
n = len(revenues)
observed_mean = revenues.mean()
s = np.std(revenues, ddof=1) 
SE = s / np.sqrt(n) 
df = n - 1
alpha = 0.05 
t_critical = st.t.ppf(1 - alpha / 2, df)
ME = t_critical * SE 
lower_bound = observed_mean - ME 
upper_bound = observed_mean + ME 

print(f"Sample mean: {observed_mean}")
print(f"t critical: {t_critical}")
print(f"Margin of error: {ME}")
print(f"95% CI: ({lower_bound}, {upper_bound})")
print(f"95% CI: ({lower_bound}, {upper_bound}), does not include 20000.")
print(f"It match with rejected H0, the true mean is somewhere between ({lower_bound}, {upper_bound})")
print("If we repeatedly took sampels of the same size and built a 95% confidence interval from each sample using the same method, about 95% of those intervals would contain the true population mean.")
```
