# Task 2 — Variance manually
We'll use the easier Hotel A first:

```python
hotel_a = [14800, 15200, 15000, 14900, 15100]
```

```python
hotel_a_std_vars = 0 
for value in hotel_a:
    hotel_a_std_vars += (value - mean_a) ** 2
population_variance = hotel_a_std_vars / len(hotel_a)
population_standard_deviation = population_variance ** 0.5
print(f"Population standard deviation is: {population_standard_deviation:.2f}")
print(f"Population variance is {population_variance:.2f}")
```

### Why do you think we square the deviations before averaging them?

1. With square the deviations we eliminate the negative numbers.
2. With square we penaltise the standard deviation from the mean. That means if the variance is big, with square we make it bigger with keeping the trend.

Imagine these are all five revenue days that exist:

```python
revenues = [14800, 15200, 15000, 14900, 15100]
# But suppose these five days were randomly selected from 365 days of revenue.
# Now they are a sample, not the population.
# Population → divide by N
# Sample     → divide by n - 1
```

Your task

Using the same hotel_a data:

2. Calculate the sample standard deviation manually.
3. Compare it with the population standard deviation of 141.42.
4. Is sample standard deviation larger or smaller?
5. Think about this without searching: why do you think we divide by n - 1 instead of n when we only have a sample?

```python
sample_variance_hotel_a = hotel_a_std_vars / (len(hotel_a)-1) 
sample_hotel_a_standard_deviation = sample_variance_hotel_a ** 0.5
print(f"1. Sample hotel_a variance is {sample_variance_hotel_a:.2f}")
print(f"2. Sample hotel_a standard deviation is: {sample_hotel_a_standard_deviation:.2f}")
print(f'3. {sample_hotel_a_standard_deviation > population_standard_deviation}')
print('4. Sample standard deviation is greater than population standard deviation.')
print('5. There is uncertainty because we only know the sample mean, not the true population mean. Dividing by n−1 instead of n corrects for the tendency of a sample to underestimate the population variance.')
```

### ***Bassel's correction is : we calculate the mean of the sample from its values so this mean sits naturally close to them. Therefore the distnace from sample mean to any value is : xi​−xˉ tend to be a little smaller than the distances from the unknown population mean would be. If simply divide by n we will tend to underestimate the population variance. Dividing by n-1 instead make the estiamted variance slightly larger to correct for this tendency. This is Bassel's correction.*** 

## Entire population
→ know/use population mean μ
→ divide by N
→ σ² and σ

## Sample used to estimate population
→ use sample mean x̄
→ divide by n − 1
