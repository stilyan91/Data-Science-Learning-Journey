# Chi-square goodness-of-fit — concept task

A hotel expects bookings to be distributed equally across four channels:

Website
Booking platform
Travel agency
Phone

So under the null hypothesis, each channel should account for: 25% of bookings.
You observe 200 bookings:

observed = np.array([70, 55, 45, 30])

corresponding to: Website, Booking platform, Travel agency, Phone

### Part A — Hypotheses

Write:

H0: 

H1: 

Think in terms of whether the observed distribution matches the expected distribution.


### Part B — Expected counts

If bookings are equally distributed across four channels, calculate the expected count for each channel.

Part C — Chi-square statistic

Use:

χ2=∑(O−E)**2/E 

where:

O = observed count
E = expected count

Calculate each category's contribution first, then sum them.

### Part D — Degrees of freedom

For goodness-of-fit:

df=k−1
where k is the number of categories.



```python
import numpy as np 
import scipy.stats as st 

observed = np.array([70,55,45,30])
k = len(observed)
print("HO: Bookings follow the expected 25%-25%-25%-25% distribution")
print("H1: Bookings do not follow that distribution.")
E = 200*0.25
xi2 = sum([(O - E)**2/E for O in observed])
print(f"chi square statistic is {xi2}")
df = k - 1 
print(f"degrees of freedom = {df}")
p_value = 1 - st.chi2.cdf(xi2, df) # p_value = st.chi2.sf(xi2, df)
print(f"P value is smaller than 0.05 {p_value} < 0.05, therefore we reject H0.")
print("There is statistically significant evidence that bookings do not follow the expected equal 25%-25%-25%-25% distribution across the channels.")
```

# Chi-square test of independence

Now we move from one categorical variable to two categorical variables.

Observed counts:

| Channel          | Cancelled | Not Cancelled |
| ---------------- | --------: | ------------: |
| Website          |        30 |           120 |
| Booking platform |        50 |           100 |
| Travel agency    |        20 |            80 |

Represent it as:
observed = np.array([
    [30, 120],
    [50, 100],
    [20,  80]
])

### Part A

Write:

H0:

H1:

Think in terms of independence/dependence between the two categorical variables

### Part B

Calculate:

- row totals
- column totals
- grand total

Then calculate the expected count for every cell using:

Eij = (row total)(colum total) / grand total
### Part C

Calculate:

xi2 = sum((O-E)**2 / E)

### Part D

Calculate degrees of freedom:

df=(r−1)(c−1)
Do not use st.chi2_contingency() yet.


```python
observed = np.array([
    [30, 120],
    [50, 100],
    [20,  80]
])

print(f"H0: booking channels and cancellation status are not related.")
print(f"H1: booking channels and cancellation status are related.")

row_totals = observed.sum(axis=1)
column_totals = observed.sum(axis=0)
grand_total = observed.sum()

r = len(observed)
c = len(observed[0])
# expected = np.array([sum(observed[idx]*sum(observed)[j]/ grand_total )  for idx in range(len(observed)) for j in range(len(observed[idx]))])
expected = np.outer(row_totals, column_totals) / grand_total
chi2 = np.sum((observed-expected)**2 / expected)
df = (r-1)*(c-1) 
# One conceptual point worth keeping: in a chi-square test of independence, the expected counts are what we would expect if H0 
# were true, meaning if the two categorical variables had no relationship.
p_value = 1-st.chi2.cdf(chi2, df)
print(f"p_value {p_value} is less than 0.05, so we reject H0.")
print("There is statistically significant evidence of an association between booking channel and cancellation status.")
st.chi2_contingency(observed)
```
