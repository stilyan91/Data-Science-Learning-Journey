# Task 8 — Bayes with a Rare Event from above
he difficulty here is that fraud is rare, so a positive flag does not automatically mean the booking is probably fraudulent.

Imagine the hotel has an automated fraud-detection system. Every booking goes through it, and the system produces either:

 
'+' → flagged as suspicious

'−' → not flagged

But the system isn't perfect.

P(F)=0.02 of all bookings are actually fraudulent.

Given that a booking really is fraudulent, there is a 90% probability the system flags it.

P(+∣F)=0.90

Given that a booking is NOT fraudulent, there is still a 5% probability the system incorrectly flags it.

P(+∣¬F)=0.05

Calculate these first:

1. P(¬F)
2. P(F∩+)
3. P(¬F∩+)
4. P(+)

```python
p_F = 0.02
p_flagged_F = 0.90
p_flagged_not_F = 0.05
p_not_F = 1 - p_F
p_F_AND_flagged = p_flagged_F * p_F 
p_not_F_AND_flagged = p_flagged_not_F * p_not_F
p_flagged = p_F_AND_flagged + p_not_F_AND_flagged
print(f"P(¬F)={p_not_F:.3f}")
print(f"Probability that a booking is fraudulent AND gets flagged is: {p_F_AND_flagged:.3f}")
print(f"Probability that a booking is NOT fraudulent AND gets flagged is: {p_not_F_AND_flagged:.3f}")
print(f"Overall probability that a random booking gets flagged is {p_flagged:.3f}")
print(f"P(F∣+)= {p_F_AND_flagged/p_flagged:.3f} Among all bookings that were flagged, about 27% are actually fraudulent.") # Among all bookings that were flagged, what proportion are actually fraudulent?
```

```python
p_R = 0.5
p_B = 0.3 
p_V = 0.2 
p_S_R = 0.2
p_S_B = 0.4
p_S_V = 0.7

p_R_AND_S = p_S_R * p_R
p_B_AND_S = p_S_B * p_B
p_V_AND_S = p_S_V *p_V
p_S = p_R_AND_S + p_B_AND_S + p_V_AND_S
p_V_S = p_S_V * p_V / p_S 

print(f"Probability some guest to be regular and to use room service is {p_R_AND_S:.3f}")
print(f"Probability some guest to be business and to use room service is {p_B_AND_S:.3f}")
print(f"Probability some guest to be VIP and to use room service is {p_V_AND_S:.3f}")
print(f"Probability some guest to use room service is {p_S:.3f}")
print(f"Given that a guest is ordered room service, probability to be VIP is {p_V_S:.3f}")
print(f"P(V∣S) is not equal to P(V), so V and s are {"Independent" if p_V_S==p_V else "Dependent"}")
```
