# Choose the test and perform it

A hotel chain wants to compare customer ratings between two independent hotels.

hotel_a = np.array([4.2, 4.4, 4.1, 4.5, 4.3, 4.0, 4.6, 4.2, 4.4, 4.3])

hotel_b = np.array([3.9, 4.1, 4.0, 4.2, 3.8, 4.1, 4.0, 3.9, 4.2, 4.0])

The question is:

Is the average customer rating different between Hotel A and Hotel B?

### Part A

Before calculating anything, write:

1. Which statistical test should be used?
2. H0 and H1.
3. Is it one-tailed or two-tailed?
4. Why is this not a paired t-test?

### Part B

Calculate for each hotel:

- n
- mean
- sample standard deviation

Then compare the two sample means informally.

### Part C

Perform the appropriate test with SciPy.

For now, use the version that does not assume equal population variances.

Store the result and print:

test statistic
p-value

Use:

α=0.05

### Part D — Interpretation

Write the conclusion in plain English.

Be careful not to say:

“The hotels definitely have different true means.”

Use statistical-evidence wording instead.

```python
import numpy as np 
import scipy.stats as st 

hotel_a = np.array([4.2, 4.4, 4.1, 4.5, 4.3, 4.0, 4.6, 4.2, 4.4, 4.3])

hotel_b = np.array([3.9, 4.1, 4.0, 4.2, 3.8, 4.1, 4.0, 3.9, 4.2, 4.0])
# part A
print("I would use two sample independent t-test.")
print(f"H0: {chr(956)}A == {chr(956)}B")
print(f"H1: {chr(956)}A != {chr(956)}B")
print('It is two-tailed test, because we ask whether the means are different in either direction.')
print('This is not a paired t-test, because the two samples are independent and hotels are two different hotels.')
# part B
nA = len(hotel_a)
nB = len(hotel_b) 
mean_A = hotel_a.mean()
mean_B = hotel_b.mean()
sA = np.std(hotel_a, ddof=1) 
sB = np.std(hotel_b, ddof=1) 
observed_mean_difference = mean_A - mean_B 
print(f"Hotel_A's sample has length {nA}, sample mean {mean_A} and standard deviation {sA}.")
print(f"Hotel_B's sample has length {nB}, sample mean {mean_B} and standard deviation {sB}.")
print(f"Informally means difference is {observed_mean_difference}")
# part C  
alpha = 0.05 
SE = np.sqrt(sA**2/nA + sB**2/nB) 
print(f"The estiamted standard error of the difference between means is {SE}.")
t = (observed_mean_difference - 0) / SE 
print(f"The observed difference between hotel_A and hotel_B is {observed_mean_difference}, that is {t} times standard error from hypothesized difference 0.")
df = (sA**2/nA + sB**2/nB) ** 2 / ((sA**2/nA)**2/(nA-1)+(sB**2/nB)**2/(nB-1))
print(f"Degrees of freedom is {df}")
p_value = 2 * (1 - st.t.cdf(abs(t), df))
print(f"p value is less than alpha {p_value} < {alpha}, so we reject H0.")
# part D 
print(f"There is statistically significant evidence that the two hotels have different true mean ratings.")
st.ttest_ind(hotel_a, hotel_b, equal_var=False)
```

before = np.array([7.1, 6.8, 7.5, 7.0, 6.9, 7.2, 7.4, 6.7, 7.3, 7.0])

Verify with SciPy using.



```python
before = np.array([7.1, 6.8, 7.5, 7.0, 6.9, 7.2, 7.4, 6.7, 7.3, 7.0])

after = np.array([7.6, 7.1, 7.8, 7.4, 7.2, 7.5, 7.7, 7.0, 7.6, 7.3])

d = after - before 
# Part A
print(f"I would use paired t-test.")
print(f"H0: {chr(956)}d <= 0")
print(f"H1: {chr(956)}d > 0")
print(f"right-tailed test, because we care about increasing in satisfacation.")
print("Paired t-test, not independet two-sample test because we measured twice the same 10 guests.")
# Part B 
differences = after - before
n = len(differences)
mean = differences.mean()
sd = np.std(differences, ddof=1) 
print(f"Length of differences is {n}, mean is {mean} and standard deviation of differences is {sd}.")
#Part C 
SEd = sd / np.sqrt(n)
print(f"Standard error of mean paired differences is about {SEd}.")
t = (mean - 0) / SEd 
print(f"Observed difference between after and before improvement is about {t} standard errors above the null-hypothesized difference 0.")
df = n -1 
print(f"Degrees of freedom is {df}.")
alpha = 0.05
#Part D 
p_value = 1 - st.t.cdf(t, df)
print(f"p value {p_value} is less than alpha {alpha}, so we reject H0.")
print(f"There is statistically significant evidence that there is increasing in satisfaction after service improvement.")
# Part E 
st.ttest_rel(after, before,alternative='greater')
```

A hotel expects booking channels to follow this distribution:

50% Direct,30% Booking.com,20% Other

observed = np.array([118, 52, 30])

corresponding to:

Use:

Then explain why this is a chi-square goodness-of-fit test and not a chi-square independence test.

### Part B — Expected counts

Calculate the expected counts from the expected proportions.

You should obtain three expected counts whose total is also 200.

### Part C — Manual chi-square statistic

Calculate:

χ2=∑(O−E)**2 / E 

Do it with NumPy.

Then calculate:

df=k−1

where k is the number of categories.

### Part D — p-value and decision

Calculate the right-tail p-value using the chi-square distribution.

Then decide whether to:

reject H0 or fail to reject H0

### Part E — Verify with SciPy


### Part F — Interpretation

Answer:

1. Is there statistically significant evidence that the observed booking-channel distribution differs from 50%-30%-20%?
2. Which category contributes most to the chi-square statistic
3. Why does chi-square use only the right tail?

```python
observed = np.array([118, 52, 30])
k = len(observed)
alpha = 0.05 
# part A 
print(f"H0: booking channels have following 50%-30%-20% distribution.")
print(f"H1: booking channels does NOT have such distribution.")
print(f"Chi square goodnes of fit test, because we test distribution between channels is as we expect.")
# part B 
expected = np.array([0.5* 200, 0.3*200, 0.2*200])
total = expected.sum() 
# part C 
chi2 = sum((observed-expected)**2/expected)
df = k - 1 
print(f"Chi-square statistic = {chi2}, degree of freedom = {df}.")
# part D 
p_value = st.chi2.sf(chi2, df)
print(f"P value is less than alpha {p_value} < 0.05, so we reject H0.")
# part E 
st.chisquare(observed,expected)
# part F 
print(f"There is statistically significant evidence that the observed booking-channel distribution differs from 50%-30%-20%.")
print(f"Direct category contributes most to the chi-square statistic.")
print("Chi square tests use only right-tailed, because the rejection region is in the right tail and larger chi2 means larger disagreement with H0.")
```

# Chi-square Test of Independence

A hotel wants to know whether booking channel is associated with booking status.

Observed counts:

observed = np.array([
    [70, 20],   # Direct: Completed, Cancelled
    [50, 30],   # Booking.com
    [30, 20]    # Other
])
Rows: Direct, Booking.com, Other
Columns: Completed, Cancelled

Use:

α=0.05

### Part A — Hypotheses

Write: H0 and H1.

Then explain why this is a chi-square test of independence and not a goodness-of-fit test.

### Part B — Expected counts manually

Calculate:

row totals
column totals
grand total

Then for every cell:

Eij = (row total)(col total) / grand total

Build the full expected-count matrix.

### Part C — Manual chi-square statistic

Calculate:

χ2 =∑(E(O−E)**2/E)

Then calculate:

df=(r−1)(c−1)

where r is number of rows and c number of columns.

### Part D — p-value and decision

Calculate the right-tail p-value using the chi-square distribution.

Then decide whether to:

reject H0 or fail to reject H0

### Part E — Verify with SciPy
Compare:

- chi-square statistic
- p-value
- degrees of freedom
- expected counts

### Part F - Interpretation 
Answer:

1. Is there statistically significant evidence that booking channel and booking status are associated?
2. Which cell contributes most to the chi-square statistic?
3. What would it mean if observed and expected counts were almost identical?

```python
observed = np.array([
    [70, 20],   # Direct: Completed, Cancelled
    [50, 30],   # Booking.com
    [30, 20]    # Other
])
alpha = 0.05 
# Part A
print(f"H0: Booking channels are NOT associated with booking status.")
print(f"H1: Booking channels are associated with booking status.")
print("We use chi-square test of independence, because we test if booking channel is associated with booking status.")
# Part B
row_total = observed.sum(axis=1)
col_total = observed.sum(axis=0)
grand_total = observed.sum()
E = np.outer(row_total,col_total)/grand_total
print(f"Expected values: \n {E}")
# Part C  
chi2 = np.sum((observed-E)**2/E)
r = len(observed)
c = len(observed[0])
df = (r-1)*(c-1)
print(f"Chi-square statistic = {chi2}, degrees of freedom = {df}.")
# Part D 
p_value = st.chi2.sf(chi2, df) 
print(f"p value is less than alpha {p_value} < 0.05, so we reject H0.")
# Part E 
st.chi2_contingency(observed)
# Part F 
print(f"There is statistically significant evidence that booking channel is associated with booking status.")
print("The Direct–Cancelled cell contributes the most to the chi-square statistic.")
print("If observed and expected counts were almost identical, that would indicate little evidence of association between the variables.")
```
