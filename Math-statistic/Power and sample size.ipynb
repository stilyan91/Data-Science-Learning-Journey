# Taks 1 
A hotel tests whether a new pricing strategy increases average daily revenue.

Use:

H0:μ≤20,000
H1:μ>20,000

Assume:

σ=3000
α=0.05

We will compare two sample sizes:

n=25

and

n=100

Assume the true mean is actually:

μ=20,800

### Part A — Standard errors

Calculate the SE for both sample sizes:

SE = sigma / sqrt(n)

Then answer:

1. Which sample size has the smaller SE?
2. By what factor does SE change when n goes from 25 to 100?
3. Which sample size should give higher statistical power, and why?

```python
import numpy as np 
import scipy.stats as st 
sigma = 3000 
n1 = 25 
n2 = 100 

se_25 = sigma / np.sqrt(n1) 
se_100 = sigma / np.sqrt(n2) 

print(f"SE_25 = {se_25}")
print(f"SE_100 = {se_100}")

print("Sample with larger size has smaller standard error.")
print("Standard error has changed by factor 2 when n goes from 25 to 100.")
print("A larger sample size reduce the standard error, so sampling distribution becomes narrower. That makes a real difference between μ and μ0 eisier to detect, which lowers beta and increase power.")
```

### Part B — Critical sample mean
Because this is a right-tailed test with:

α=0.05

For each sample size, calculate the critical sample mean:
x_critical = m0 + z_critical * SE 

Then answer:

1. What sample mean is needed to reject H0 when n=25?
2. What sample mean is needed to reject H0 when n=100?
3. Why does the larger sample make it easier to detect the true mean of £20,800?


```python
alpha = 0.05
z_critical = st.norm.ppf(1 - alpha)
m0 = 20000
x_critical_25 = m0 + z_critical * se_25 
x_critical_100 = m0 + z_critical * se_100 

print(f"The sample mean beyond which we reject H0 for sample size 25 is {x_critical_25}")
print(f"The sample mean beyond which we reject H0 for sample size 100 is {x_critical_100}.")
print("A larger sample reduce the standard error and distribution becomes narrower. That makes a real difference between m and m0 eisier to detect.")

```

### Part C: Calculate β and Power
Now we use the true mean:

μ1=20,800

and ask:

If the true mean really is £20,800, what is the probability that we still fail to reject H0?

That probability is:

β=P(Type II error)

For this right-tailed test, we fail to reject H0 when:

xˉ≤xˉcrit

So under the true distribution centered at μ1=20800:

β=P(Xˉ≤xˉcrit∣μ=20800)

Step 1 — Standardize the critical sample mean

For each sample size, calculate: 

zβ=(xˉcrit−μ1) / SE 

Step 2 — Convert to probability

Because β is the probability to the left of the critical sample mean


Step 3 — Calculate power

Power=1−β

Then answer:

1. What is β for n=25?
2. What is power for n=25?
3. What is β for n=100?
4. What is power for n=100?
5. Which sample size has much higher power, and why?

```python
m1 =20800

z_beta_25 = (x_critical_25 - m1) / se_25
z_beta_100 = (x_critical_100 - m1) / se_100 

beta_25 = st.norm.cdf(z_beta_25)
beta_100 = st.norm.cdf(z_beta_100)

power_25 = 1 - beta_25
power_100 = 1 - beta_100

print(f"beta for n = 25 is {beta_25}")
print(f"Power for n = 25 is {power_25}")
print(f"beta for n = 100 is {beta_100}")
print(f"Power for n = 100 is {power_100}")
print(f"Sample size 100 has much power, because standard error is are lower, which makes true shift from 20000 to 20800 eisier to detect\
and reduce the probability of type II error.")
# One important interpretation:

#power =0.38 means only about a 38% chance to detect the real increase when n=25
#power =0.85 means about an 85% chance to detect it when n=100
# n goes up => SE goes down => beta goes down => Power goes UP.
```
