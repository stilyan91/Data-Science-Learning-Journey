Scenario — Hotel guests

A hotel has 2,500 guests.

- 1,500 booked Online.
- 1,000 booked through a Travel Agency.
- 900 guests purchased Breakfast.
- 600 of the Online guests purchased Breakfast.
- Every guest used exactly one booking channel.

Let:

O=Online
T=Travel Agency
B=Breakfast
Part A — Basic probabilities

Calculate:

1. P(O)
2. P(T)
3. P(B)
4. P(O∩B)
### Part B — Conditional probability

Calculate:

5. P(B∣O)
6. P(O∣B)

Then explain in words what the difference is between those two probabilities.

### Part C — Independence
Determine whether Online and Breakfast are independent or dependent.

Prove your conclusion using both methods:

P(B∣O)
=
?
P(B)

and

P(O∩B)
=
?
P(O)P(B)
### Part D — Reasoning

Without being directly given the number, determine:

8. How many Travel Agency guests purchased Breakfast?

Then calculate:

9. P(B∣T)
10. Which group is more likely to purchase breakfast: Online or Travel Agency guests?

```python
guests = 2500
online_booked = 1500
travel_agency_booked = 1000
guest_had_breakfast  = 900
breakfast_online_booked = 600

# use:  wanted event / all possible outcome  
p_O = online_booked / guests
p_T = travel_agency_booked / guests
p_B = guest_had_breakfast / guests
# here calculate what part of guests who booked online, have purchased breakfast.
p_O_AND_B = breakfast_online_booked / guests
# P(B∣O) if guest is online, what is probability to order breakfast 
p_B_O = breakfast_online_booked / online_booked
# P(O∣B) if guest purchased a breakfast, what is probability to come from online platform.
p_O_B = breakfast_online_booked / guest_had_breakfast
# Travel agency guest ordered breakfast.
T_B = guest_had_breakfast-breakfast_online_booked
p_B_AND_T = T_B / guests
# P(B∣T)
p_B_T = p_B_AND_T / p_T 

print(f"Probability a guest to booked online is {p_O:.3f}")
print(f"Probability a guest to booked through travel agency is {p_T :.3f}")
print(f"Probability a guest to purchesed breakfast is {p_B :.3f}")
print(f"Probability a guest to booked online and have breakfast is {p_O_AND_B:.3f}")
print(f"If guest has booked online, the probability to order breakfast is {p_B_O:.3f}")
print(f"If guest purchased breakfast, the probability to come from online is {p_O_B:.3f}")
print(f"Test 1: {"Independent" if p_B_O==p_B else "Dependent"}")
print(f"Test 2: {"Independent" if p_O_AND_B == p_O * p_B else "Dependent"}")
print("Both test shows that breakfast and online are Dependent.")
print(f"{T_B} Travel agency guest ordered breakfast.")
print(f"Probability a guest booked by travel agency to order breakfast is {p_B_T:.3f}")
print(f'P(B∣O)({p_B_O})>P(B∣T)({p_B_T}) shows that in this dataset online guests are more likely to order breakfast.')
```

### Part D — Reasoning without calculation first

Before writing Python, answer these:

7. A booking is known to be cancelled. Which denominator should you use to calculate the probability that it was Mobile: 3000, 1800, or 480? Why?

8. Which cancellation rate is higher: Mobile or Desktop?

9. Does that automatically mean that most cancelled bookings came from the channel with the higher cancellation rate? Explain why the size of each channel could matter.

Question 9 is the new idea. A rate within a group and a share of all cancelled bookings are different things.

```python
print("Denominator will be 480, because we we are looking for answer only in cancelled bookings.")
print("Mobile has more cancellation rate because it has 3 times more cancelled bookings in 1.5 times more total bookings.")
print("No, does not automatically mean most cancelled bookings come from channel with higher cancellation rate, because some times \
very small share of something big is bigger than significant part from very small thing.")
print("A higher rate within a subgroup does not necessary")

total_bookings = 3000 
mobile = 1800
desktop = 1200 
cancelled = 480
completed = 2520

p_M = mobile / total_bookings
p_C = cancelled / total_bookings
p_M_AND_C = 360 / total_bookings
p_C_M = 360 / mobile
p_M_C = 360 / cancelled
p_C_D = 120 / desktop
p_M_OR_C = p_M + p_C - p_M_AND_C
print(f"Probability of mobile booking is {p_M:.3f}")# we search only in total bookings
print(f"Probability of cancelled booking is {p_C:.3f}")#  we search only in total bookings
print(f"Probability of mobile and cancelled booking is {p_M_AND_C:.3f}") # we search only in total bookings
print(f'Probability to have mobile booking and to be cancelled is {p_C_M:.3f}')# we search only in mobile bookings
print(f'{p_M_C:.3f} among cancelled bookings, come from mobile booking')# we search only in cancelled bookings
print(f'{p_C_D:.3f} among desktop bookings, are cancelled.') # we search only in desktop bookings
print()
print(f"{'Independent' if p_C_M  == p_C  else "Dependent"}")
print(f"{'Independent' if p_M_AND_C== p_M * p_C else "Dependent"}")
print("Both test prove that mobile and cancelled are dependent.")
print()
print(f'P(M∪C) = {p_M_OR_C:.3f}')
```

```python
print("If 75 % of all Spa users are Loyal Members does NOT necessarily mean 75 % of Loaylty Members use spa because those are different events.\
P(S|L) != P(L|S). First One is among loyalty members what portion used spa and other event is what portion among spa users are loyalty members.")
print()
total_guests = 4000
loyalty_members = 2400
non_members = 1600
guest_used_spa = 1200
loyalty_members_spa = 900

p_L = loyalty_members / total_guests
p_S = guest_used_spa / total_guests 
p_L_AND_S = loyalty_members_spa / total_guests
p_S_L = p_L_AND_S / p_L
p_L_S = p_S_L * p_L / p_S 
# first calc intersection of spa and non-member guests 
NM_guest_used_spa = guest_used_spa - loyalty_members_spa
p_S_NM = NM_guest_used_spa / non_members
print(f"P(L) = {p_L:.3f}") # because we search what part from total geusts, are loalty members. 
print(f"P(S) = {p_S:.3f}") # because we search what part from total geusts, are spa users. 
print(f"P(L∩S) = {p_L_AND_S:.3f}") # What proportion of all guests are both Loyalty Members and Spa users?
print(f"P(S∣L)= {p_S_L:.3f}") # numerator - Intersection of the loalty_members and spa users, denominator - probability to be loalty member.
print(f"P(L∣S)= {p_L_S:.3f}") # bayes theorem 
print()
print(f"{"Independet" if p_S_L == p_S else "Dependent"}") # Knowing that a guest is a Loyalty Member does change the probability of Spa.
print(f"{"Independet" if p_L_AND_S == p_L*p_S else "Dependent"}") # intersection of loalty members and spa users can't be calculated like independent probability.
print("Both tests for checking indenpendence failed, that means Loalty member and spa are dependent probabilities.")
print(f"{NM_guest_used_spa} non-members used SPA")
print(f"Calculate P(S∣NonMember) = {p_S_NM:.3f}")
print(f"Calculate P(S∣NonMember) < P(S∣L) {p_S_NM*100:.2f}% < {p_S_L*100:.2f}%, ")
print(f"Loalty members group uses more often spa by {(p_S_L-p_S_NM)*100} percentage points.")
print()
print("P(L∩S) - what part of total users are loyalty member and used spa.")
print("P(L∣S) - knowing that guest used spa, what is probabilty to be loaylty member") 
```
