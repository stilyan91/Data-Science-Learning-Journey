# Task 1 — Understanding a Normal Distribution

Suppose we collect the daily revenue of a hotel over many days.

The revenue is approximately normally distributed with:

μ=£20,000

and:

σ=£2,000

where:

μ = population mean
σ = population standard deviation

A normal distribution is centered around the mean and is symmetric:

                         mean
                          ↓
                       £20,000
                          |
                    .-----------.
                 .-'             '-.
               .'                   '.
             .'                       '.
____________/___________________________\____________
        £16k       £18k       £20k       £22k       £24k
          -2σ        -1σ         μ          +1σ        +2σ
Remember what standard deviation tells us: it describes how far values typically spread from the mean.

Here:

1σ=£2,000

So £22,000 is one standard deviation above the mean, while £18,000 is one standard deviation below it.

Your questions

1. Is £20,000 below, above, or exactly at the mean?

2. How many standard deviations above the mean is £22,000?

3. How many standard deviations below the mean is £16,000?

4. A day has revenue of £21,000. Without using a formula yet, is it:

less than 1 standard deviation above the mean,
exactly 1 standard deviation above,
or more than 1 standard deviation above?

5. Reasoning: Two days have revenues:

A=£18,000
B=£24,000

Which day's revenue is farther from the mean in terms of standard deviations? Explain.
          

```python
print("1. Exactly at the mean.")
print("2. One time standard deviation above the mean.")
print("3. Two time standard deviation less than the mean.")
print("4. 21 000£ is less than 1 standard deviation above the mean.")
print("5. 24 000£ is farther from mean in term of standard deviation, because 24 000 - 20 000 is 4000, that is 2 times std deviation from mean.")
```

# Task 2: Introducing the Z-score

We already know how to think about values as a number of standard deviations from the mean. A z-score simply gives that distance a number.

The formula is:
z=σx−μ

where:

x = observation
μ = mean
σ = standard deviation
z = number of standard deviations from the mean

Using our hotel revenue example:

μ=£20,000,σ=£2,000

### Part A — Calculate the z-scores

Calculate the z-score for each revenue:

1. x=£20,000
2. x=£22,000
3. x=£18,000
4. x=£25,000
5. x=£15,000

Use Python and calculate them yourself with the formula.

### Part B — Interpret them

For each result, explain what it means.

For example, don't just write:

z=1

Write:

The revenue is 1 standard deviation above the mean.

Pay attention to the sign:

z>0⇒above mean
z<0⇒below mean
z=0⇒at mean

### Part C — Reasoning

6. Without calculating additional z-scores: Which is more unusual: £25,000 or £18,000? Why?
7. Two different hotels have:

Hotel A

μA=£20,000,σA=£2,000

Today's revenue:

xA=£24,000

Hotel B

μB=£50,000,σB=£10,000

Today's revenue:

xB=£60,000

Which hotel had the more unusually high revenue relative to its normal performance?


```python
mean = 20_000
std = 2_000
x = [20000, 22000, 18000, 25000, 15000]
for i in range(len(x)): 
    # globals()[f'x_{i}'] = {(x[i] - mean)/std}
    z = (x[i] - mean)/std
    print(f"The revenue is {z} standard deviation {'above the' if z > 0 else 'below the' if z<0 else 'at'} mean.")

print("£25,000 is more than usual because it is 2.5 std deviation above the mean, till £18,000 is -1 std deviation below the mean.")
print("Hotel A because xA​=£24,000 is 2 std deviation above the mean, xB​=£60,000 is 1 time std deviation below the mean.")
```

Don't calculate z-scores yet. First build the visualization yourself.

Look at your histogram itself.

Would you say these 15 revenue observations provide enough evidence to confidently claim:

"Revenue is normally distributed"

Why or why not?

Don't calculate anything. Think about sample size and the shape of the histogram first.

```python
import numpy as np 
import matplotlib.pyplot as plt 

revenues = np.array([
    18000, 19000, 19500, 20000, 20500,
    21000, 22000, 18500, 19800, 20200,
    21500, 20500, 19200, 20800, 20000
])
mean = revenues.mean()
std_dev = revenues.std()
plt.hist(revenues,)
plt.axvline(mean,linewidth=2,color='r', label='Mean')
plt.axvline(mean-std_dev,linewidth=2,color='g', label='-1 std Dev')
plt.axvline(mean+std_dev,linewidth=2, color='b', label='+1 std Dev')
plt.title('Distribution of the Revenues')
plt.xlabel("Revenues")
plt.ylabel("Count")
plt.legend()
plt.show()
revenues.shape
print("The region from μ−σ to μ+σ contains observations that are within one standard deviation of the mean.")
print("Those 15 revenue observations are NOT enough to confidenply claim 'Revenue is normally distributed.'")
```

revenues = [

### Part C — Visualize

Create another histogram and add vertical lines for:

Then additionally mark the observation that is farthest from the mean with another vertical line.

### Part D — Interpretation

Suppose your most extreme observation has:

z=2.1

Explain in plain English what 2.1 means.

```python
from scipy import stats as st 

revenues = np.array([
    18000, 19000, 19500, 20000, 20500,
    21000, 22000, 18500, 19800, 20200,
    21500, 20500, 19200, 20800, 20000
])
z_scores = st.zscore(revenues)

for revenue,z_score in zip(revenues, z_scores):
    print(f'Revenue: £{revenue} -> z = {z_score.round(2)}')

print("EXtremes")
print("Largest positive z-score is ", z_scores.max())
print("Smallest/most negative z-score is ", z_scores.min())
idx = np.argmax(np.abs(z_scores))

print(f"The farthest from mean overall is {revenues[idx]}")

plt.hist(revenues)
plt.axvline(revenues.mean(), linewidth=1,color='r',label='Mean')
plt.axvline(revenues.mean()+revenues.std(), linewidth=1,color='b', label="+1 Std Dev")
plt.axvline(revenues.mean()-revenues.std(), linewidth=1, color='b', label="-1 Std Dev")
plt.axvline(revenues[idx], linewidth=1, color='g', label='Farthest Observation')
plt.title('Distribution of Revenues')
plt.xlabel('Revenues')
plt.ylabel('Count')
plt.legend(loc='upper right')
plt.show()

print("z=2.1 means that observation values it 2.1 standard deviation above the mean.")
```
