A hotel receives bookings through three channels:
| Channel              | % of all bookings | Cancellation rate |
| -------------------- | ----------------: | ----------------: |
| Website (W)          |               50% |               10% |
| Booking Platform (B) |               30% |               20% |
| Travel Agency (T)    |               20% |               25% |

Let C = booking is cancelled.

###  Part A — AND & conditional probability

Calculate:

1. P(W∩C)
2. P(B∩C)
3. P(T∩C)

Choose the appropriate multiplication rule yourself.

### Part B — Total probability
Calculate the overall probability that a randomly selected booking is cancelled:

4. P(C)
Calculate:

5. P(Not Cancelled)

### Part C — Bayes

You now know that a particular booking was cancelled.

Calculate:

6. P(W∣C)
7. P(B∣C)
8. P(T∣C)

Then verify something interesting:

P(W∣C)+P(B∣C)+P(T∣C)=?

Explain why you get that result.

### Part D — Independence
Are W and C independent?

Prove your answer using either:

P(C∣W)=?P(C)

or:

P(W∣C)=?P(W)

### Part E — Reasoning challenge
Without calculating anything new, answer:

A cancelled booking is selected. Is it more likely to have come from Website or Booking Platform?

Be careful here. Website has a lower cancellation rate, but it also generates many more bookings.

Explain your reasoning using the probabilities you already calculated.

```python
print("Without calculation I would say Platform will have more cancelled booking.")

p_W_AND_C = 0.1 * 0.5
p_B_AND_C = 0.2 * 0.3
p_T_AND_C = 0.25 * 0.2

p_C = p_W_AND_C + p_B_AND_C + p_T_AND_C
p_NOT_C = 1 - p_C

p_W_C = p_W_AND_C / p_C
p_B_C = p_B_AND_C / p_C
p_T_C = p_T_AND_C / p_C


print(f"Probability booking to come from WEBSITE and to be cancelled is {p_W_AND_C:.2f}")
print(f"Probability booking to come from PLATFORM and to be cancelled is {p_B_AND_C:.2f}")
print(f"Probability booking to come from TRAVEL AGENCY and to be cancelled is {p_T_AND_C:.2f}")
print(f"The overall probability that a randomly selected booking is cancelled {p_C:.3f}")
print(f"The overall probability that a randomly selected booking is NOT cancelled {p_NOT_C:.3f}")
print(f"P(W∣C) = {p_W_C:.3f}")
print(f"P(B∣C) = {p_B_C:.3f}")
print(f"P(T∣C)) = {p_T_C:.3f}")
print(f"P(W∣C)+P(B∣C)+P(T∣C)={p_W_C+p_B_C+p_T_C}, because W,B and T are all available options to cancel a booking.")
print(f"W and C are {"Independent" if p_W_C== 0.5 else "Dependent"}, because P(W∣C) is not equal to P(W)")
```
