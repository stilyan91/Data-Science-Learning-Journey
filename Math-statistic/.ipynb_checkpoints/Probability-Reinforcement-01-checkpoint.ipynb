# Task 1 — Basic probability and complements
A hotel has 2,000 bookings:

Status	Bookings
- Completed	1,420
- Cancelled	380
- No-show	200

Total	2,000

Calculate:

1. P(Completed)
2. P(Cancelled)
3. P(NoShow)
4. P(Not Cancelled)
5. P(Not Completed)

Then answer:

If I tell you that P(Cancelled)=0.19, can you calculate P(Completed) using only that information? Why or why not?

```python
completed = 1420
cancelled = 380 
no_show = 200 
total = 2000
p_completed = completed/total 
p_cancelled = cancelled/total 
p_no_show = no_show/total 
p_not_cancelled = p_completed + p_no_show
p_not_completed = p_cancelled + p_no_show
print(f"1. Probability of the completed bookings is {p_completed:.2f} or {p_completed*100}%")
print(f"1. Probability of the cancelled bookings is {p_cancelled:.2f} or {p_cancelled*100}%")
print(f"1. Probability of the No show bookings is {p_no_show:.2f} or {p_no_show*100}%")
print(f"1. Probability of the Not cancelled bookings is {p_not_cancelled:.2f} or {p_not_cancelled*100}%")
print(f"1. Probability of the Not completed bookings is {p_not_completed:.2f} or {p_not_completed*100:.1f}%")

print("I can't caculate P(Completed) only with information about P(cancelled), because there is an option no show. \
If options were only 2: completed and cancelled, I would use rule of complement 1-P(cancelled) and will  get P(completed)")
```

# Task 5 — Choose the Formula Yourself

A hotel reports:

P(B)=0.35

where B = guest orders breakfast.

And:

P(S)=0.20

where S = guest uses the spa.

We also know:

P(S∣B)=0.30
Calculate:
1. P(B∩S)
2. Are Breakfast and Spa independent or dependent?
3. P(B∪S)

Then suppose:

P(L∣S)=0.60

where L = Loyalty Member.

And:

P(S)=0.20,P(L)=0.30

4. Calculate P(S∣L).

For question 4, think about the formula you asked me about earlier. This is the first time in this reinforcement practice where Bayes' theorem is genuinely useful.

```python
p_breakfast = 0.35
p_spa = 0.20
p_spa_given_breakfast = 0.30
# p(S|B) = P(B∩S) / P(B) => P(B∩S) = P(B)*p(S|B) 
p_breakfast_and_spa = p_breakfast * p_spa_given_breakfast 
print(f"{p_breakfast_and_spa:.2f} I choosed that formula, because they are dependent events") 
print(f"P(S|B) = 0.30 != P(S) = 0.20 that means knowing that event breakfast is happened, the probability for spa is changed. Therefore those \
two events are dependent.")
p_breakfast_or_spa = p_breakfast + p_spa - p_breakfast_and_spa
print(f"{p_breakfast_or_spa:.2f} I choosed this formula, because those two events don't overlap.")

# p(S|L) = p(L\S) * p(S) / p(L) Bayes' theorem.
print(f"P(S|L) = {0.60*0.20/0.3:.2f} ")
```

# Multi-Step Probability

A hotel chain has two booking channels:

60% of bookings are made through Mobile
40% are made through Desktop

Cancellation rates are:

25% of Mobile bookings are cancelled
10% of Desktop bookings are cancelled

Let:

M=Mobile
D=Desktop
C=Cancelled
Calculate
1. P(M∩C)
    Probability that a booking is Mobile AND Cancelled
2. P(D∩C)
     Probability that a booking is Desktop AND Cancelled
3. Overall probability that a randomly selected booking is cancelled:
P(C)

Think about the fact that cancellations can come from either Mobile or Desktop.

4. Probability that a cancelled booking came from Mobile:
P(M∣C)

For this one, Bayes' theorem will be useful.

Interpretation

5. Which channel has the higher cancellation rate?
6. Among all cancelled bookings, do you expect most cancellations to come from Mobile or Desktop? Explain why before calculating if possible.
7. Is the fact that Mobile has more cancellations necessarily evidence that Mobile itself is worse? Think about the fact that Mobile also has more total bookings.

```python
p_mobile = 0.6
p_desktop = 0.4
p_mobile_calcellation_rate = 0.25
p_desktop_cancellation_rate = 0.10

p_mobile_and_cancelled = p_mobile * p_mobile_calcellation_rate
p_desktop_and_cancelled = p_desktop * p_desktop_cancellation_rate
p_cancelled = p_mobile_and_cancelled + p_desktop_and_cancelled
p_mobile_given_cancelled = p_mobile_and_cancelled / p_cancelled
print(f"Probability that booking is Mobile and Cancelled is: {p_mobile_and_cancelled:.2f}")
print(f"Probability that booking is Desktop and Cancelled is: {p_desktop_and_cancelled:.2f}")
print(f"Probability that a random selected booking is cancelled: {p_cancelled:.2f}")
print(f"Probability that cancelled booking came from Mobile is: {p_mobile_given_cancelled:.2f}")
print(f"Mobile channel has higher cancellation rate.")
print("I expected the most cancellation to come from Mobile because it has more bookings and also has biggier cancellation rate.")
print("Mobile has both more bookings and a higher cancellation rate, so its larger number of cancellations is not caused only by having more bookings. The cancellation rate itself is worse.")
```

### Calculate:

```python
p_breakfast = 0.4 
p_spa = 0.3
p_breakfast_and_spa = 0.18

p_breakfast_or_spa = p_breakfast + p_spa - p_breakfast_and_spa
p_spa_given_breakfast = p_breakfast_and_spa / p_breakfast
p_breakfast_given_spa = p_spa_given_breakfast * p_breakfast / p_spa
print(f"Probability for guest to order breakfast or spa is: {p_breakfast_or_spa:.2f}") # i choose formula for overlapping adition, because 1 guest can order breakfast and use spa similteniously
print(f"Probability guest to order breakfast and after that to use spa is: {p_spa_given_breakfast:.2f}") # Formula for conditional rule, because we know intersection of those events.
print(f"Probability of guest who had breakfast and used spa is: {p_breakfast_given_spa:.2f}") #I used bayes theorem because already know probabilty for reverse happening of events.
print()
print("Independent" if p_spa_given_breakfast == p_spa else "Dependent") # first check for dependency between those probabilities.
print("Independent" if p_breakfast_and_spa == p_spa*p_breakfast else "Dependent") # second check for dependency between those probabilities.
print("After 2 test, it was proven that both guest to use spa and order breakfast are DEPENDENT. First check shows that happening of breakfast increase probability to use spa.\
and second shows that guest use breakfast and spa could NOT be calculate by formula for independent events.")
print()
p_l = 0.25
p_b_l = 0.60
p_breakfast_and_loyal_member = p_b_l * p_l 
p_breakfast = 0.40
p_loyal_breakfast = p_b_l * p_l/p_breakfast
print(f"Probability a guest to be loayal member and order breakfast is {p_breakfast_and_loyal_member:.2f}")
print(f"Probability a guest to order breakfast and to be loyal member is: {p_loyal_breakfast:.2f}") # Bayes theorem.
print()
print("6. It tells me that those probabilities are dependent.")
print("7. It tells me that those probabilities are Independent.")
print("8. If P(A∣B)=0 and P(B) > 0, therefore P(A∣B)=P(A∩B)/P(B)=0 ->P(A∩B) = 0, so A and B cannot occur together. Therefore they are mutually exclusive.")
```

```python
p_mobile = 0.70
p_desktop = 0.30

p_cancelled_mobile = 0.20
p_cancelled_desktop = 0.10

p_not_cancelled_mobile = 1 - p_cancelled_mobile
p_not_cancelled_desktop = 1 - p_cancelled_desktop

# I multiply the probabilities when follow the tree path, because each second-stage probability is conditional on the first-stage event.
p_mobile_and_cancelled = p_cancelled_mobile * p_mobile
p_mobile_and_not_cancelled = p_not_cancelled_mobile * p_mobile

p_desktop_and_cancelled = p_cancelled_desktop * p_desktop
p_desktop_and_not_cancelled = p_not_cancelled_desktop*p_desktop
print(f"Probability of booking made by mobile and not to be cancelled is: {p_not_cancelled_mobile:.2f}")
print(f"Probability of booking made by desktop and not to be cancelled is: {p_not_cancelled_desktop:.2f}")
print(f"Probability of the booking to be mobile and cancelled is {p_mobile_and_cancelled:.2f}")
print(f"Probability of the booking to be mobile and NOT cancelled is {p_mobile_and_not_cancelled:.2f}")
print(f"Probability of the booking to be desktop and cancelled is {p_desktop_and_cancelled:.2f}")
print(f"Probability of the booking to be desktop and NOT cancelled is {p_desktop_and_not_cancelled:.2f}")
print()
sum_probability = p_mobile_and_cancelled + p_mobile_and_not_cancelled +p_desktop_and_cancelled+p_desktop_and_not_cancelled 
print(f"Sum of all four probabilities is {sum_probability}, because it is all posible outcomes for 1 bookings. Probability can be value from 0-1")
print()
p_cancelled = p_mobile_and_cancelled + p_desktop_and_cancelled 
print(f"The overall probability of cancelletion is {p_cancelled:.2f}, we add both probabilities because we cannot cancel the booking both way.\
we choose mobile or desktop. They are mutually exclusive.")
p_mobile_cancelled = p_mobile_and_cancelled / p_cancelled
print(f"Probability of one randome cancelled booking to be from mobile is {p_mobile_cancelled:.2f}")
print("P(Cancelled∣Mobile) - the booking is made by mobile, what is probability to be cancelled.")
print("P(Mobile∣Cancelled) - the booking is cancelled, what is probability to be made by mobile.")
```

```python
total = 2000
direct = 1200 
platform = 800

p_breakfast_direct = .55
p_breakfast_platform = .30

p_D = direct/total
p_P = platform/total
p_D_AND_B = p_breakfast_direct * p_D
p_P_AND_B = p_breakfast_platform * p_P
p_B = p_D_AND_B + p_P_AND_B
p_D_B = p_D_AND_B / p_B 
p_D_OR_B = p_D + p_B - p_D_AND_B
print(f"Probability of guest to book by Direct channel is {p_D:.3f}")
print(f"Probability of guest to book by Platform channel is {p_P:.3f}")
print(f"Probability of direct guest order breakfast is {p_breakfast_direct}")
print(f"Probability of platform guest order breakfast is {p_breakfast_platform:.2f}")
print(f"Probability of guest ot order breakfast and be direct is {p_D_AND_B:.3f}")
print(f"Probability of guest ot order breakfast and be Platform is {p_P_AND_B:.3f}")
print(f"Probability of guest ot order breakfast is {p_B:.3f}")
print(f"Probability of guest who has already breafast to be direct is {p_D_B:.3f}")
print(f"P(D∪B)={p_D_OR_B:.3f} represents represents probability guest to be direct OR to have breakfast.")
print(f"Test 1: Direct and breakfast are {'Independet' if p_breakfast_direct == p_B  else 'Depdendent'}")
print("Test 1 tells me that happening of event direct, changes the probability of event breakfast.")
print(f"Test 2: Direct and breakfast are {'Independet' if  p_D_AND_B == p_D * p_B  else 'Depdendent'}")
print("Test 2 shows that the intersection cannot be calculated as P(D)P(B), because the events are not independent.")

print("P(B∣D) is probability to direct guest to have breakfast.")
print("P(D∣B) is probability to guest who have breakfast, to be direct.")
print("30% of guests those come from platform order breakfast, but it does not mean the most guests do not order breakfast came from platform, \
because Direct guest have more total size bookings.")

print("If guests next mount are 10 000 P(B) would NOT change because every proportions will be kept. We would expect more breakfast orders in absolute terms but still 45 % of guests.")

print("P(not A)= 1 - P(A)")
print("P(A∪B)= p(A) + p(B) - p(A ∩ B) - if they overlap. else P(A∪B)= p(A) + p(B)")
print("P(A∣B)= P(A∩B) / p(B)")
print('P(A∩B)= P(A∣B) * p(B)') # general
print('P(A∩B)= p(A) * p(B)') # independent
print("P(A∣B)= P((B|A) * P(A) / P(B)") # Bayes
```
