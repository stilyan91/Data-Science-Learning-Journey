# Mixed Exercise 1

For each scenario, identify:

which statistical test to use
H0 and H1

whether the test is one-sided/two-sided where applicable
### A

A hotel chain claims its average daily revenue is £20,000. You take a sample of 25 days and want to test whether the true mean is different.

### B

You want to compare average revenue between Hotel A and Hotel B. The observations come from different hotels and are independent.

### C

You measure the same 20 hotels’ occupancy before and after a new pricing strategy.

### D

A hotel expects bookings to come from these channels in proportions:

50% Direct,30% Booking.com,20% Other

You collect actual booking counts and want to test whether the observed proportions match those expected proportions.

### E

You record:

booking channel: Direct / Booking.com / Other
booking status: Completed / Cancelled

and want to know whether booking status is associated with booking channel.


```python
print('''
Scenario A
Test: one-sample t-test 
H0: true mean is 20000
H1: true mean is different from 20000.
Direction: two-tailed test, because we are looking if true mean is 20k.
----------------------
Scenario B
Test: two sample independent t-test
H0: hotel_A's mean is equal to hotel_B's mean
H1: hotel_A's mean is different to hotel_B's mean
Direction: two-tailed test, we care about changes in both directions.
----------------------
Scenario C
Test: paired sample t-test
H0: after - before = 0 
H1: after - before != 0
Direction: two-tailed test
----------------------
Scenario D
Test: chi-square goodness of fit 
H0: bookings following 50%-30%-20% distribution between channels. 
H1: bookings does not following such distribution.
Direction: right-tailed test, the rejection region is in right tail and larger χ2 means larger disagreement with H0.
----------------------
Scenario E
Test: chi-square independence test
H0: bookings Booking channels and booking status are not related. 
H1: bookings Booking channels and booking status are related.
Direction: right-tailed test, the rejection region is in right tail and larger χ2 means larger disagreement with H0.
----------------------
''')
```
