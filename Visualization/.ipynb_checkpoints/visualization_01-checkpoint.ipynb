# Task 8 — Histogram

```python
import numpy as np 

np.random.seed(42)

revenue = np.random.normal(
    loc=15000,
    scale=2500,
    size=500
)
```

Create four separate histograms using:
- bins = 5
- bins = 10
- bins = 30
- bins = 100

Then answer:

- What happens to the histogram when you increase the number of bins?
- Which number of bins gives you the clearest picture of the distribution?
- Where does most of the revenue appear to be concentrated?
- Does the distribution look symmetric, left-skewed, or right-skewed?
- What do you think loc=15000 and scale=2500 mean in np.random.normal()?

Don't look up question 5 yet. Infer it from the histogram and the numbers. This exercise starts preparing you for the normal distribution → standard deviation → CLT → z-test progression.


```python
bins = [5,10,30,100]

fig, axes = plt.subplots(2,2,figsize=(12,8))
axes_flat = axes.flatten()
for i in range(len(bins)):
    axes_flat[i].hist(revenue,bins=bins[i],edgecolor='black')
    axes_flat[i].set_title(f'Bins: {bins[i]}')
    axes_flat[i].set_ylabel('Counts')
    axes_flat[i].set_xlabel("Value")
plt.tight_layout()
plt.show()
```

1. With increase the number of bins, the histogram loses its symmetric
2. With 30 bins, it is the clearest picture of distribution
3. Most of the revenue is concentrated from 12000 to 17500.
4. The distribution looks like symmetric.
5. I think loc is where the most value will be concentrated, scale is how the data will be dispersed. 

# Task 9 — Box Plot

```python
revenue = [
    12000, 12500, 12800, 13000, 13200,
    13500, 13800, 14000, 14200, 14500,
    14800, 15000, 15500, 16000, 35000
]
```

Create a box plot with Matplotlib.

Then, without calculating everything with pandas, answer:

- Approximately where is the median?
- Is there an obvious outlier?
- Which value is it?
- What do you think the box itself represents?
- Remember our earlier 1.5 × IQR exercise: what do you think the whiskers represent?

```python
plt.figure(figsize=(12,8))
plt.boxplot(revenue)
plt.title('Box Plot of the Revenue')
plt.ylabel("Revenue (GBP)")
plt.tight_layout()
plt.show()
```

1. Median is arround 14000.
2. Yes, there is an outlier.
3. The value of outlier is 35000
4. Box represent the distribution of the data.
5. Whiskers are the area under lower bound q1 - 1.5 * IQR till q3 + 1.5*IQR
